# Exploration d'un dataset réel : Titanic

## Contexte

Dans cet exercice, je vais explorer le dataset Titanic avec Pandas.

Le dataset contient des informations sur les passagers du Titanic :
- leur classe ;
- leur âge ;
- leur sexe ;
- le prix du billet ;
- s'ils ont survécu ou non ;
- le nombre de frères/sœurs ou conjoints à bord ;
- le nombre de parents/enfants à bord.

L'objectif est de découvrir les données et de répondre à 10 questions
à l'aide de Pandas.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("titanic.csv")

df.head()
```

## Question 1 — Quelle est la taille du dataset ?

Je commence par regarder le nombre de lignes et de colonnes du dataset.

La méthode `shape` retourne un tuple :
- le premier élément correspond au nombre de lignes ;
- le deuxième correspond au nombre de colonnes.

```python
df.shape
```

Le dataset contient donc **891 passagers** et **12 colonnes**.

## Question 2 — Quels sont les types des colonnes ?

Je regarde maintenant le type de chaque colonne avec `dtypes`.

Cela permet de savoir si une colonne contient des nombres,
du texte ou des valeurs booléennes.

```python
df.dtypes
```

Les colonnes numériques sont principalement `PassengerId`, `Survived`,
`Pclass`, `Age`, `SibSp`, `Parch` et `Fare`.

Les colonnes comme `Name`, `Sex`, `Ticket`, `Cabin` et `Embarked`
contiennent du texte.

## Question 3 — Y a-t-il des valeurs manquantes ?

Je vais compter le nombre de valeurs manquantes dans chaque colonne.

Pour cela, j'utilise `isna()` puis `sum()`.

```python
df.isna().sum()
```

Les principales valeurs manquantes se trouvent dans :

- `Cabin` : 687 valeurs manquantes ;
- `Age` : 177 valeurs manquantes ;
- `Embarked` : 2 valeurs manquantes.

La colonne `Cabin` est particulièrement incomplète.

## Question 4 — Quelles sont les statistiques descriptives ?

Je vais utiliser `describe()` pour obtenir un résumé statistique
des colonnes numériques.

Cela permet notamment de voir :
- le nombre de valeurs ;
- la moyenne ;
- l'écart-type ;
- le minimum ;
- les quartiles ;
- le maximum.

```python
df.describe()
```

Cette analyse permet d'avoir une première idée de la distribution
des variables numériques.

Par exemple, on peut observer que le tarif moyen (`Fare`) est
d'environ 32,20 et que l'âge moyen des passagers connus est
d'environ 29,7 ans.

## Question 5 — Quelle est la corrélation entre les variables numériques ?

Je vais calculer la matrice de corrélation des variables numériques.

La corrélation permet de voir si deux variables ont tendance
à évoluer ensemble.

Une valeur proche de 1 indique une corrélation positive forte,
une valeur proche de -1 une corrélation négative forte,
et une valeur proche de 0 une faible corrélation linéaire.

```python
correlations = df.corr(numeric_only=True)

correlations.round(2)
```

On peut notamment observer une corrélation négative entre `Pclass`
et `Fare`.

Cela signifie que les passagers des classes les plus élevées
(numériquement plus faibles : 1ère classe = 1) ont tendance à avoir
des tarifs plus élevés.

On observe également une relation entre `Pclass` et `Survived`.

## Question 6 — Quels passagers ont survécu ?

La colonne `Survived` indique si le passager a survécu :

- `0` = non ;
- `1` = oui.

Je vais filtrer le dataset pour récupérer uniquement les passagers
qui ont survécu.

```python
survivants = df[df["Survived"] == 1]

survivants.head()

survivants.shape
```

Il y a donc **342 passagers survivants** dans le dataset.

## Question 7 — Quels passagers avaient plus de 60 ans ?

Je vais utiliser un filtre Pandas pour sélectionner les passagers
dont l'âge est supérieur à 60 ans.

Les valeurs manquantes de `Age` ne seront pas sélectionnées.

```python
passagers_plus_60 = df[df["Age"] > 60]

passagers_plus_60[["Name", "Age", "Sex", "Pclass", "Survived"]]
```

Ce filtre permet de se concentrer uniquement sur les passagers
les plus âgés et de comparer ensuite leur classe, leur sexe
et leur survie.

## Question 8 — Quels sont les passagers avec les billets les plus chers ?

Je vais trier le dataset selon la colonne `Fare`.

J'utilise `sort_values()` avec `ascending=False` afin d'obtenir
les tarifs les plus élevés en premier.

```python
df_trie_fare = df.sort_values(
    by="Fare",
    ascending=False
)

df_trie_fare[
    ["Name", "Pclass", "Sex", "Age", "Fare", "Survived"]
].head(10)
```

J'obtiens ainsi les 10 passagers ayant payé les billets les plus chers.

Le tri permet de repérer rapidement les valeurs les plus élevées
de la colonne `Fare`.

## Question 9 — Quel est le nombre de passagers par classe ?

Je vais utiliser `groupby()` pour regrouper les passagers
selon leur classe (`Pclass`).

Ensuite, je compte le nombre de passagers dans chaque groupe.

```python
passagers_par_classe = df.groupby("Pclass").size()

passagers_par_classe
```

La troisième classe est la plus représentée avec **491 passagers**.

La première classe compte **216 passagers** et la deuxième
**184 passagers**.

## Question 10 — Quel est le taux de survie selon le sexe ?

Pour terminer, je vais utiliser `groupby()` pour comparer
la survie des hommes et des femmes.

Comme `Survived` contient 0 et 1, calculer la moyenne de cette
colonne donne directement le taux de survie.

```python
taux_survie_sexe = df.groupby("Sex")["Survived"].mean()

taux_survie_sexe = (
    df.groupby("Sex")["Survived"]
    .mean()
    .mul(100)
    .round(2)
)

taux_survie_sexe
```

Le taux de survie est beaucoup plus élevé chez les femmes.

Environ **74,20 % des femmes ont survécu**, contre seulement
**18,89 % des hommes**.

Cette différence est l'un des éléments les plus visibles
dans l'exploration du dataset Titanic.
