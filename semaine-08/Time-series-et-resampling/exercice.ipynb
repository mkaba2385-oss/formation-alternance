# Time series et resampling — Bitcoin

Dans cet exercice, on manipule des données temporelles du Bitcoin.

Objectifs :
- convertir la colonne `date` en index datetime ;
- calculer une moyenne mobile sur 7 jours ;
- calculer une moyenne mobile sur 30 jours ;
- calculer le retour quotidien en % ;
- calculer la volatilité mensuelle avec un resampling ;
- visualiser le prix et les moyennes mobiles.

## 1. Importation des bibliothèques

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Chargement du dataset

Le fichier CSV doit s'appeler `bitcoin_daily.csv` et être placé dans le même dossier que ce notebook.

```python
df = pd.read_csv("bitcoin_daily.csv")

df.head()
```

## 3. Vérification des données

```python
df.info()
df.describe()
```

## 4. Conversion de la date en datetime

La colonne `date` est convertie en type datetime pour pouvoir effectuer des opérations temporelles.

```python
df["date"] = pd.to_datetime(df["date"])

df.dtypes
```

## 5. Mettre la date comme index

On utilise la date comme index afin de pouvoir exploiter les fonctionnalités de pandas dédiées aux séries temporelles.

```python
df = df.set_index("date")
df = df.sort_index()

df.head()
```

## 6. Moyenne mobile sur 7 jours

La moyenne mobile permet de lisser les variations du prix. Ici, chaque valeur correspond à la moyenne des 7 derniers jours.

```python
df["ma_7"] = df["price"].rolling(window=7).mean()

df[["price", "ma_7"]].head(10)
```

## 7. Moyenne mobile sur 30 jours

Même principe que précédemment, mais avec une fenêtre de 30 jours.

```python
df["ma_30"] = df["price"].rolling(window=30).mean()

df[["price", "ma_7", "ma_30"]].head(35)
```

## 8. Retour quotidien en %

`pct_change()` calcule la variation du prix par rapport au jour précédent.

```python
df["daily_return"] = df["price"].pct_change() * 100

df[["price", "daily_return"]].head(10)
```

## 9. Volatilité mensuelle

On utilise le resampling pour passer des rendements quotidiens à une fréquence mensuelle. La volatilité est calculée avec l'écart-type des rendements quotidiens de chaque mois.

```python
monthly_volatility = df["daily_return"].resample("ME").std()

monthly_volatility = monthly_volatility.to_frame(name="monthly_volatility")

monthly_volatility.head()
```

## 10. Résumé des données calculées

```python
df.head(40)
```

## 11. Graphique du prix et des moyennes mobiles

On superpose le prix du Bitcoin, la moyenne mobile à 7 jours et la moyenne mobile à 30 jours.

```python
plt.figure(figsize=(14, 7))

plt.plot(df.index, df["price"], label="Prix Bitcoin")
plt.plot(df.index, df["ma_7"], label="Moyenne mobile 7 jours")
plt.plot(df.index, df["ma_30"], label="Moyenne mobile 30 jours")

plt.title("Bitcoin - Prix et moyennes mobiles")
plt.xlabel("Date")
plt.ylabel("Prix")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

## 12. Graphique de la volatilité mensuelle

Cette visualisation permet de voir les mois pendant lesquels les rendements du Bitcoin ont été les plus variables.

```python
plt.figure(figsize=(14, 5))

plt.plot(
    monthly_volatility.index,
    monthly_volatility["monthly_volatility"],
    marker="o",
)

plt.title("Volatilité mensuelle du Bitcoin")
plt.xlabel("Mois")
plt.ylabel("Volatilité (%)")
plt.grid(True)
plt.tight_layout()

plt.show()
```

Dans cet exercice, j'ai manipulé une série temporelle contenant les prix quotidiens du Bitcoin.

J'ai commencé par convertir la colonne `date` en `DatetimeIndex`. Cela permet à pandas de reconnaître les données comme une série temporelle.

J'ai ensuite calculé deux moyennes mobiles, une sur 7 jours et une sur 30 jours, afin de lisser les variations du prix.

Le retour quotidien en pourcentage permet de mesurer l'évolution du prix d'un jour à l'autre.

Enfin, j'ai utilisé le resampling mensuel pour calculer la volatilité à partir des rendements quotidiens.
